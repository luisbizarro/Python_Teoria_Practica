## EJERCICIO 11

![problema11](https://i.imgur.com/ZYqAhLN.png)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Definir la función
def f3(x):
    return (4*x - 2)**(1/2)

x = np.linspace(0.5, 20.5, 400)

# Crear un rango de valores x
y = f3(x)

# Graficar la función
plt.figure()
plt.plot(x, y, label='f(x) = (4x-2)^1/2)', color='blue')
plt.title('Gráfica de la Función Racional')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.axhline(0, color='red', lw=0.5, ls='--')
plt.axvline(0, color='green', lw=0.5, ls='--')
plt.grid()
plt.legend()
plt.show()
```
